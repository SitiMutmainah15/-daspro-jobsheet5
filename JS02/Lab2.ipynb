```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('Titanic-Dataset.csv')
df.head()
```

```python
# Age - mean
df['Age'] = df['Age'].fillna(df['Age'].mean())

# Cabin - DECK
df['Cabin'] = df['Cabin'].fillna("DECK")

# Embarked - modus
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.to_csv('Titanic-Dataset-Fixed.csv', index=False)
```
